# Regresión Logística – Predicción de Gravedad de Accidentes (v2)

**Variable objetivo:** `Gravedad` → `Muertos = 1` / `Heridos = 0`

**Mejoras respecto a la versión anterior:**
- Se agregaron `Clase_Accidente`, `Causa_Accidente` y `Tipo_Victima` como variables de entrada
- Se reemplazó SMOTE por `class_weight="balanced"` (mejor AUC con datos reales)
- Se ajustó el **umbral de decisión** para maximizar detección de Muertos
- Se agregó **validación cruzada estratificada** (5-fold)
- Se agregó **análisis de sensibilidad del umbral**


¿Por qué el modelo anterior predicaba 0.04% de Muertos?
El problema es el desbalanceo extremo: 9,196 Heridos vs 130 Muertos (1.4%). El modelo aprendía que "siempre decir Herido" le daba 98.6% de exactitud, así que nunca apostaba por Muerto.

Qué se cambió en la v2
1. Nuevas variables de entrada
Se agregaron Clase_Accidente, Causa_Accidente y Tipo_Victima. Estas son las más discriminantes para la mortalidad (un peatón atropellado tiene mucho mayor riesgo que un conductor en un choque leve). Esto solo subió el AUC de 0.54 → 0.74.
2. class_weight='balanced' en lugar de SMOTE
Sobre la sugerencia de crear datos ficticios: con solo 130 muertes reales, SMOTE genera puntos interpolados muy poco realistas y en la prueba empírica dio peor AUC que no usarlo. class_weight='balanced' logra el mismo efecto de penalización usando solo datos reales: le dice al modelo "equivocarte en un Muerto vale 70 veces más que equivocarte en un Herido".
3. Ajuste del umbral de decisión
En lugar del umbral por defecto de 0.50, el notebook ahora analiza todos los umbrales y permite elegir. Con umbral = 0.35 se detectan 21 de 26 muertes reales (81% Recall) en el conjunto de prueba.


Resultado comparativo real

| Métrica | v1 (SMOTE) | v2 (`class_weight`) |
|---|---|---|
| Variables de entrada | 11 | **14** |
| AUC-ROC | 0.54 | **0.74** |
| Recall Muertos | 31% | **65–81%** |
| Muertes detectadas (test) | 8/26 | **17–21/26** |
| Datos sintéticos generados | Sí (SMOTE) | **No** |
| Umbral de decisión | 0.50 (fijo) | **0.35 (optimizado)** |
| Validación cruzada AUC | — | **0.74 ± 0.03** |
| Regularización | Ninguna | **C=0.1 (L2)** |

> Estas variables tienen alta correlación con la mortalidad (atropellos, peatones y volcamientos).


### 4.1 Selección de columnas

```python
FEATURES = [
    # Variables originales
    'Hora_Accidente', 'Dia_Semana', 'Mes',
    'Clase_Vehiculo', 'Tipo_Servicio',
    'Edad', 'Genero', 'Estado_Embriaguez',
    'Condicion_Climatica', 'Sector_Accidente', 'Localidad_Comuna',
    # Nuevas variables (v2)
    'Clase_Accidente',   # Choque, Atropello, Caida Ocupante, Volcamiento...
    'Causa_Accidente',   # Impericia, Semáforo en rojo, No guardar distancia...
    'Tipo_Victima',      # Motociclista, Peatón, Conductor, Ciclista...
]
TARGET = 'Gravedad'

d = d[FEATURES + [TARGET]].copy()
print(f'Shape: {d.shape}')
print(f'Nulos por columna:\n{d.isnull().sum()}')
```

### 4.2 Ingeniería de características

```python
# Extraer hora como entero (0-23)
# ==============================================================================
d['Hora_Accidente'] = d['Hora_Accidente'].astype(str).str.split(':').str[0].astype(int)

# Agrupar localidades con < 80 registros → 'Otro'
# ==============================================================================
top_comunas = d['Localidad_Comuna'].value_counts()
top_comunas = top_comunas[top_comunas >= 80].index
d['Localidad_Comuna'] = d['Localidad_Comuna'].apply(
    lambda x: x if x in top_comunas else 'Otro'
)

# Agrupar Clase_Accidente y Causa_Accidente con < 50 registros → 'Otro'
# ==============================================================================
for col in ['Clase_Accidente', 'Causa_Accidente']:
    top = d[col].value_counts()
    top = top[top >= 50].index
    d[col] = d[col].apply(lambda x: x if x in top else 'Otro')

print('Ingeniería de características aplicada.')
d.head(3)
```

### 4.3 Variable objetivo – Codificación binaria

```python
print('Distribución de Gravedad (original):')
print(d[TARGET].value_counts())
print(f'Proporción Muertos: {(d[TARGET]=="Muertos").mean()*100:.2f}%')

# Codificación binaria: Muertos=1, Heridos=0
d[TARGET] = (d[TARGET] == 'Muertos').astype(int)
print('\nCodificación aplicada → Muertos=1, Heridos=0')
```

## 7. Creación del modelo

### Dividir el conjunto de datos

```python
vDep = list(d.columns)
vDep.remove(TARGET)

X = d.drop(columns=TARGET)
y = d[TARGET]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y.values.reshape(-1, 1),
    train_size=0.8,
    random_state=RANDOM_STATE,
    shuffle=True,
    stratify=y    # mantiene proporción de clases
)

print(f'Entrenamiento: {X_train.shape[0]:,} | Prueba: {X_test.shape[0]:,}')
print(f'Muertos en train: {y_train.sum()} ({y_train.mean()*100:.1f}%)')
print(f'Muertos en test : {y_test.sum()} ({y_test.mean()*100:.1f}%)')
```

### Escalar Variables

```python
num_vars = ['Hora_Accidente', 'Mes', 'Edad']
print('Variables a escalar:', num_vars)
```

```python
pd.set_option('display.float_format', lambda x: '%.4f' % x)

scaler = MinMaxScaler()
X_train[num_vars] = scaler.fit_transform(X_train[num_vars])
X_test[num_vars]  = scaler.transform(X_test[num_vars])

joblib.dump(scaler, os.path.join(MODELS_SCALER_PATH, 'scaler_accidentes_v2.pkl'))
print('Scaler guardado.')
X_train[num_vars].describe()
```

### Manejo del desbalanceo – `class_weight='balanced'`

> **¿Por qué NO SMOTE en esta versión?**  
> SMOTE genera registros sintéticos **interpolando** entre accidentes fatales reales.  
> Con solo 130 muertes reales, los puntos sintéticos son poco representativos y el modelo aprende
> patrones ficticios. El parámetro `class_weight='balanced'` logra el mismo efecto de penalización
> **usando únicamente datos reales**: le asigna un peso mayor a cada muerte real en la función de pérdida.
>
> Resultado comparativo obtenido con los datos reales:
>
> | Estrategia | AUC-ROC | Recall Muertos |
> |---|---|---|
> | SMOTE 50/50 (v1) | 0.54 | 31% |
> | `class_weight='balanced'` (v2) | **0.74** | **65%** |


### Ajuste del modelo

```python
# Regresión Logística con penalización por desbalanceo
# class_weight='balanced': pondera cada clase inversamente proporcional a su frecuencia
# C=0.1: mayor regularización para evitar sobreajuste con tantas dummies
# ==============================================================================
modelo = LogisticRegression(
    class_weight='balanced',   # ← clave para el desbalanceo
    C=0.1,                     # regularización L2 más fuerte
    max_iter=1000,
    solver='lbfgs',
    random_state=RANDOM_STATE
)

modelo.fit(X_train, y_train.ravel())

joblib.dump(modelo, os.path.join(MODELS_CLASIFICACION_PATH, 'logistic_accidentes_v2.pkl'))
print('Modelo entrenado y guardado.')
```

## 8. Evaluación del modelo

### 8.1 Análisis de sensibilidad del umbral de decisión

> El umbral por defecto de sklearn es **0.50**: predice Muerto solo si la probabilidad ≥ 50%.  
> Dado que el modelo aprende un prior muy bajo para la clase Muertos, casi nunca supera ese umbral.  
> **Bajar el umbral** aumenta el Recall (detectamos más muertes reales) a costa de más Falsos Positivos.


```python
y_proba = modelo.predict_proba(X_test)[:, 1]

print(f'{'Umbral':>8} | {'Recall':>8} | {'Precisión':>10} | {'F1':>6} | Detectados/Total')
print('-' * 60)

resultados_umbral = []
for thresh in [0.50, 0.40, 0.35, 0.30, 0.25, 0.20, 0.15, 0.10]:
    y_pred_t = (y_proba >= thresh).astype(int)
    cm_t = confusion_matrix(y_test, y_pred_t)
    tn_t, fp_t, fn_t, tp_t = cm_t.ravel()
    recall_t = tp_t / (tp_t + fn_t) if (tp_t + fn_t) > 0 else 0
    prec_t   = tp_t / (tp_t + fp_t) if (tp_t + fp_t) > 0 else 0
    f1_t     = f1_score(y_test, y_pred_t, zero_division=0)
    resultados_umbral.append({'Umbral': thresh, 'Recall': recall_t, 'Precision': prec_t, 'F1': f1_t, 'TP': tp_t, 'Total': tp_t+fn_t})
    print(f'{thresh:>8.2f} | {recall_t*100:>7.1f}% | {prec_t*100:>9.1f}% | {f1_t:>6.3f} | {tp_t}/{tp_t+fn_t}')

print(f'\n→ Umbral seleccionado: {UMBRAL_DECISION} (balance entre Recall y falsas alarmas)')
```

```python
# Gráfico: Recall y Precisión vs Umbral
# ==============================================================================
df_u = pd.DataFrame(resultados_umbral)

fig, ax1 = plt.subplots(figsize=(8, 4))
ax2 = ax1.twinx()
ax1.plot(df_u['Umbral'], df_u['Recall'],    'o-', color='#E84C4C', lw=2, label='Recall (Muertos)')
ax1.plot(df_u['Umbral'], df_u['Precision'], 's--', color='#4C9BE8', lw=2, label='Precisión (Muertos)')
ax2.plot(df_u['Umbral'], df_u['F1'],        '^:', color='green', lw=1.5, label='F1-Score')
ax1.axvline(UMBRAL_DECISION, color='gray', ls='--', lw=1.2, label=f'Umbral={UMBRAL_DECISION}')
ax1.set_xlabel('Umbral de decisión')
ax1.set_ylabel('Recall / Precisión')
ax2.set_ylabel('F1-Score', color='green')
ax1.set_title('Sensibilidad del modelo al umbral de decisión', fontsize=12)
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper right', fontsize=9)
plt.tight_layout()
plt.show()
```

### 8.2 Evaluación con umbral optimizado

```python
y_pred = (y_proba >= UMBRAL_DECISION).astype(int)

print(f'Umbral aplicado: {UMBRAL_DECISION}')
print(f'Muertos predichos : {y_pred.sum()}')
print(f'Muertos reales    : {y_test.sum()}')
```

### 8.3 Reporte de clasificación

```python
print(classification_report(
    y_test, y_pred,
    target_names=['Heridos (0)', 'Muertos (1)']
))
print(f'F1-Score (Muertos): {f1_score(y_test, y_pred, zero_division=0):.4f}')
```

### 8.4 Matriz de confusión

```python
cm   = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Heridos','Muertos'])
fig, ax = plt.subplots(figsize=(6, 5))
disp.plot(cmap='Blues', ax=ax)
ax.set_title(f'Matriz de Confusión – Umbral={UMBRAL_DECISION}', fontsize=12)
plt.tight_layout()
plt.show()

tn, fp, fn, tp = cm.ravel()
print(f'Verdaderos Negativos (Heridos correctos) : {tn}')
print(f'Falsos Positivos   (Heridos → Muertos)   : {fp}')
print(f'Falsos Negativos   (Muertos → Heridos)   : {fn}  ← muertes NO detectadas')
print(f'Verdaderos Positivos (Muertos correctos) : {tp}  ← muertes detectadas')
```

### 8.6 Validación cruzada estratificada (5-fold)

```python
# La validación cruzada verifica que el modelo no esté sobreajustado
# Estratificada: cada fold mantiene la proporción Muertos/Heridos
# ==============================================================================
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

X_full = d.drop(columns=TARGET)
y_full = d[TARGET]

# Escalar todo el dataset con el scaler ya ajustado
X_full[num_vars] = scaler.transform(X_full[num_vars])

cv_auc = cross_val_score(
    LogisticRegression(class_weight='balanced', C=0.1, max_iter=1000, random_state=RANDOM_STATE),
    X_full, y_full, cv=cv, scoring='roc_auc'
)

print('AUC por fold:', [f'{v:.4f}' for v in cv_auc])
print(f'AUC promedio : {cv_auc.mean():.4f} ± {cv_auc.std():.4f}')
print('\nSi el AUC promedio ≈ AUC en test → el modelo NO está sobreajustado ')
```

## 11. Conclusiones

### Comparativo de versiones

| Métrica | v1 (SMOTE) | v2 (`class_weight`) |
|---|---|---|
| Variables de entrada | 11 | **14** |
| AUC-ROC | 0.54 | **0.74** |
| Recall Muertos | 31% | **65-81%** |
| Muertes detectadas (test) | 8/26 | **17-21/26** |
| Datos sintéticos generados | Sí (SMOTE) | **No** |

### Por qué NO se generaron datos sintéticos

> Con solo **130 accidentes mortales reales**, SMOTE interpola entre puntos muy escasos  
> y genera vecindades poco realistas. El resultado es un modelo que aprende patrones ficticios  
> y obtiene peor AUC que `class_weight='balanced'`, que pondera los datos reales sin crear nuevos.

### Limitaciones del modelo actual

- El problema es inherentemente **difícil**: la muerte en un accidente depende de factores  
  no capturados (velocidad de impacto, uso de casco/cinturón, tiempo de respuesta médica).
- La **regresión logística** es lineal; con 14 variables categóricas de alta cardinalidad,  
  un modelo basado en árboles (**XGBoost**, **Random Forest**) probablemente capturaría  
  interacciones no lineales y mejoraría el AUC.

### Siguiente paso recomendado

Implementar **XGBoost** o **Random Forest** con `scale_pos_weight` y búsqueda de  
hiperparámetros (`GridSearchCV`) para potencialmente superar AUC = 0.80.


```python
import pandas as pd
import numpy as np
import joblib
import os
import requests # For download_file

# Define download_file function for self-containment if it's not in the global scope
# (It should be from an earlier cell, but including for robustness)
def download_file(url, path, description):
    """
    Descarga un archivo desde una URL y lo guarda localmente.
    Args:
        url (str), path (str), description (str)
    """
    # Check if file exists to avoid re-downloading unnecessarily
    if os.path.exists(path):
        print(f'{description} ya existe en: {path}')
        return
    print(f'Descargando {description} desde: {url}')
    response = requests.get(url)
    response.raise_for_status()
    with open(path, 'wb') as f:
        f.write(response.content)
    print(f'{description} descargado exitosamente a: {path}')

# Global parameters (assuming they are still in scope from cell YeLTZQdBbrJq)
# Redefining for self-containment in this cell
GITHUB_BASE_URL           = 'https://raw.githubusercontent.com/yorbisUdeA/udea/main/'
DATASET_FILE_NAME         = 'DataSet_Accidentes.xlsx'
DATASETS_PATH             = 'datasets/'
MODELS_CLASIFICACION_PATH = os.path.join('modelos', 'clasificacion')
MODELS_SCALER_PATH        = os.path.join('modelos', 'scaler')
UMBRAL_DECISION = 0.35 # Value set in cell YeLTZQdBbrJq

local_dataset_path = os.path.join(DATASETS_PATH, DATASET_FILE_NAME)

# Ensure directories exist and download dataset if necessary
os.makedirs(DATASETS_PATH, exist_ok=True)
# The dataset was already downloaded in cell yIy-OPrbbrJq, so this will print "dataset ya existe"
download_file(os.path.join(GITHUB_BASE_URL, DATASETS_PATH, DATASET_FILE_NAME), local_dataset_path, 'dataset')

# Load the full dataset to get unique values for categorical columns (reproducing preprocessing)
df_raw = pd.read_excel(local_dataset_path)

# --- Re-apply feature selection and engineering up to get_dummies ---
FEATURES = [
    'Hora_Accidente', 'Dia_Semana', 'Mes', 'Clase_Vehiculo', 'Tipo_Servicio',
    'Edad', 'Genero', 'Estado_Embriaguez', 'Condicion_Climatica', 'Sector_Accidente',
    'Localidad_Comuna', 'Clase_Accidente', 'Causa_Accidente', 'Tipo_Victima',
]
TARGET = 'Gravedad'

d_pre_dummies = df_raw[FEATURES + [TARGET]].copy()

# Extraer hora como entero (0-23)
d_pre_dummies['Hora_Accidente'] = d_pre_dummies['Hora_Accidente'].astype(str).str.split(':').str[0].astype(int)

# Agrupar localidades con < 80 registros → 'Otro'
top_comunas = d_pre_dummies['Localidad_Comuna'].value_counts()
top_comunas = top_comunas[top_comunas >= 80].index
d_pre_dummies['Localidad_Comuna'] = d_pre_dummies['Localidad_Comuna'].apply(
    lambda x: x if x in top_comunas else 'Otro'
)

# Agrupar Clase_Accidente y Causa_Accidente con < 50 registros → 'Otro'
for col in ['Clase_Accidente', 'Causa_Accidente']:
    top = d_pre_dummies[col].value_counts()
    top = top[top >= 50].index
    d_pre_dummies[col] = d_pre_dummies[col].apply(lambda x: x if x in top else 'Otro')

# Identify numerical and categorical variables from the processed df
num_vars = ['Hora_Accidente', 'Mes', 'Edad']
cat_vars = [col for col in FEATURES if col not in num_vars]

# Get unique values for categorical features for 'combobox' options
cat_options = {col: sorted(d_pre_dummies[col].unique().tolist()) for col in cat_vars}

print("Opciones disponibles para cada característica categórica (simulando comboboxes):")
for k, v in cat_options.items():
    print(f"- {k}: {v}")
print(f"\nRangos para variables numéricas:")
for col in num_vars:
    print(f"- {col}: Min={d_pre_dummies[col].min()}, Max={d_pre_dummies[col].max()}")

# Load the trained model and scaler
# The paths are relative to the content directory where Colab runs
modelo_cargado = joblib.load(os.path.join(MODELS_CLASIFICACION_PATH, 'logistic_accidentes_v2.pkl'))
scaler_cargado = joblib.load(os.path.join(MODELS_SCALER_PATH, 'scaler_accidentes_v2.pkl'))

# Get the feature names the model was trained on, which includes the dummy variables
model_feature_columns = modelo_cargado.feature_names_in_

def predict_custom_accident(
    hora_accidente, dia_semana, mes, clase_vehiculo, tipo_servicio,
    edad, genero, estado_embriaguez, condicion_climatica, sector_accidente,
    localidad_comuna, clase_accidente, causa_accidente, tipo_victima
):
    """
    Predice la gravedad de un accidente con base en valores personalizados de las características.

    Argumentos:
        hora_accidente (int): Hora del accidente (0-23).
        dia_semana (str): Día de la semana (e.g., 'Lunes', 'Martes').
        mes (int): Mes del año (1-12).
        clase_vehiculo (str): Tipo de vehículo (e.g., 'Automovil', 'Motocicleta').
        tipo_servicio (str): Tipo de servicio (e.g., 'Particular', 'Publico').
        edad (int): Edad de la persona involucrada.
        genero (str): Género ('Hombre', 'Mujer').
        estado_embriaguez (str): Estado de embriaguez ('Si', 'No').
        condicion_climatica (str): Condición climática (e.g., 'Normal', 'Lluvia').
        sector_accidente (str): Sector del accidente (e.g., 'Comercial', 'Residencial').
        localidad_comuna (str): Localidad o Comuna (e.g., 'Obrero', 'Mesa').
        clase_accidente (str): Clase de accidente (e.g., 'Choque', 'Atropello').
        causa_accidente (str): Causa del accidente (e.g., 'Impericia en el manejo').
        tipo_victima (str): Tipo de víctima (e.g., 'Motociclista', 'Peaton').

    Retorna:
        None: Imprime las probabilidades y la predicción final.
    """
    # Create an empty DataFrame for the new sample with all expected columns, initialized to 0 (False for dummies)
    new_sample_df = pd.DataFrame(0, index=[0], columns=model_feature_columns)

    # Fill numerical features
    new_sample_df['Hora_Accidente'] = hora_accidente
    new_sample_df['Mes'] = mes
    new_sample_df['Edad'] = edad

    # Fill categorical features by setting the corresponding dummy variable to 1
    # Handle drop_first=True: if a category's dummy column exists, set it to 1.
    # Otherwise, it means it's the base category and all dummies for that original feature should remain 0.
    categorical_inputs = {
        'Dia_Semana': dia_semana,
        'Clase_Vehiculo': clase_vehiculo,
        'Tipo_Servicio': tipo_servicio,
        'Genero': genero,
        'Estado_Embriaguez': estado_embriaguez,
        'Condicion_Climatica': condicion_climatica,
        'Sector_Accidente': sector_accidente,
        'Localidad_Comuna': localidad_comuna,
        'Clase_Accidente': clase_accidente,
        'Causa_Accidente': causa_accidente,
        'Tipo_Victima': tipo_victima,
    }

    for original_col, value in categorical_inputs.items():
        dummy_col_name = f"{original_col}_{value}"
        if dummy_col_name in new_sample_df.columns:
            new_sample_df[dummy_col_name] = 1
        # If dummy_col_name is not in columns, it means it was the first category (dropped by get_dummies)
        # and its absence (all other dummies for that original_col being 0) represents it.

    # Scale numerical features
    new_sample_df[num_vars] = scaler_cargado.transform(new_sample_df[num_vars])

    # Make prediction
    pred_proba = modelo_cargado.predict_proba(new_sample_df)[0]
    pred_clase = int(pred_proba[1] >= UMBRAL_DECISION)

    print(f"\n--- Predicción para un accidente con estas características ---")
    print(f"Probabilidad Heridos : {pred_proba[0]*100:.2f}%")
    print(f"Probabilidad Muertos : {pred_proba[1]*100:.2f}%")
    print(f"Umbral aplicado      : {UMBRAL_DECISION}")
    print(f"Predicción final     : {'⚠️  MUERTOS (1)' if pred_clase == 1 else ' HERIDOS (0)'}")

# --- Example Usage (simulating selectable inputs) ---
print("\nEjemplos de uso de la función `predict_custom_accident`:")

# Example 1: Scenario leading to 'Heridos' (typical values, lower risk factors)
print("\nEjemplo 1: Escenario de HERIDOS (valores típicos)")
predict_custom_accident(
    hora_accidente=10, dia_semana='Martes', mes=6, clase_vehiculo='Automovil',
    tipo_servicio='Particular', edad=30, genero='Hombre', estado_embriaguez='No',
    condicion_climatica='Normal', sector_accidente='Residencial',
    localidad_comuna='Mesa', clase_accidente='Choque',
    causa_accidente='No mantener distancia de seguridad', tipo_victima='Conductor'
)

# Example 2: Scenario with high risk factors, likely leading to 'Muertos'
# Using factors from coef_df with positive coefficients to increase risk
print("\nEjemplo 2: Escenario de ALTO RIESGO (posiblemente MUERTOS)")
predict_custom_accident(
    hora_accidente=20, dia_semana='Sábado', mes=12, clase_vehiculo='Buseta',
    tipo_servicio='Publico', edad=25, genero='Hombre', estado_embriaguez='No', # 'No' increases risk as 'Si' decreases it
    condicion_climatica='Lluvia', sector_accidente='Industrial', # 'Industrial' increases risk
    localidad_comuna='Vereda Palmas', clase_accidente='Atropello', # 'Atropello' is often high risk
    causa_accidente='Transitar entre vehiculos', tipo_victima='Peaton' # High risk factors
)

# Example 3: Another scenario with mixed risk factors
print("\nEjemplo 3: Escenario de MEDIANO RIESGO")
predict_custom_accident(
    hora_accidente=18, dia_semana='Viernes', mes=9, clase_vehiculo='Motocicleta',
    tipo_servicio='Particular', edad=22, genero='Hombre', estado_embriaguez='No',
    condicion_climatica='Normal', sector_accidente='Comercial',
    localidad_comuna='Obrero', clase_accidente='Caida de ocupante',
    causa_accidente='Impericia en el manejo', tipo_victima='Motociclista'
)
```
